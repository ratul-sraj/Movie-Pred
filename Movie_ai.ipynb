## Combine features

### Subtask:
Create a new column in the `merged_df` by combining the selected similarity features into a single string for each movie.


**Reasoning**:
Combine the selected similarity features into a single string for each movie by defining and applying a function to create a new column.



```python
def combine_features(row):
    """Combines selected features into a single string."""
    return ' '.join(str(row[col]) for col in similarity_columns)

# Apply the function to create the new 'combined_features' column
merged_df['combined_features'] = merged_df.apply(combine_features, axis=1)

# Display the first few entries of the new 'combined_features' column
print("First few entries of the 'combined_features' column:")
display(merged_df['combined_features'].head())
```

## Vectorize features

### Subtask:
Use a vectorization technique (e.g., TF-IDF) on the combined feature column to convert the text data into numerical vectors.


**Reasoning**:
Use TF-IDF vectorization to convert the combined text features into a numerical matrix.



```python
from sklearn.feature_extraction.text import TfidfVectorizer

# Instantiate TfidfVectorizer
tfidf_vectorizer = TfidfVectorizer(stop_words='english', max_features=10000) # Using common English stop words and limiting features to manage memory

# Fit and transform the 'combined_features' column
tfidf_matrix = tfidf_vectorizer.fit_transform(merged_df['combined_features'])

# Display the shape of the resulting TF-IDF matrix
print("Shape of the TF-IDF matrix:")
display(tfidf_matrix.shape)
```

## Create similarity function

### Subtask:
Define a function that takes a movie title as input and returns the top N most similar movies based on the calculated similarity scores.


**Reasoning**:
Define the `get_similar_movies` function to find similar movies based on the cosine similarity matrix.



```python
def get_similar_movies(movie_title, num_similar=5):
    """
    Finds the top N most similar movies to a given movie title.

    Args:
        movie_title: The title of the movie.
        num_similar: The number of similar movies to return (default is 5).

    Returns:
        A list of titles of the top N similar movies, or an error message
        if the movie title is not found.
    """
    # Get the index of the movie title
    try:
        movie_index = merged_df[merged_df['primaryTitle'].str.lower() == movie_title.lower()].index[0]
    except IndexError:
        return f"Movie with title '{movie_title}' not found."

    # Get the TF-IDF vector for the movie
    movie_vector = tfidf_matrix[movie_index]

    # Calculate the cosine similarity scores between the movie vector and all other movie vectors
    similarity_scores = cosine_similarity(movie_vector, tfidf_matrix).flatten()

    # Get the indices of the top N similar movies (excluding the movie itself)
    # Use numpy.argsort to get the indices that would sort the array
    top_similar_indices = similarity_scores.argsort()[-num_similar-1:-1][::-1]


    # Get the titles of the top N similar movies
    similar_movies = merged_df['primaryTitle'].iloc[top_similar_indices].to_pylist()

    return similar_movies

# Example usage (optional, for testing the function)
similar_movies_example = get_similar_movies("Toy Story", num_similar=5)
print(f"Top 5 similar movies to Toy Story: {similar_movies_example}")
```

```python
import cupy as cp

def get_similar_movies_gpu(movie_title, gpu_merged_df, gpu_cosine_sim, num_similar=5):
    """
    Finds the top N most similar movies to a given movie title using GPU.

    Args:
        movie_title: The title of the movie.
        gpu_merged_df: The cuDF DataFrame containing movie data.
        gpu_cosine_sim: The GPU-calculated cosine similarity matrix.
        num_similar: The number of similar movies to return (default is 5).

    Returns:
        A list of titles of the top N similar movies, or an error message
        if the movie title is not found.
    """
    # Get the index of the movie title from the cuDF DataFrame
    try:
        # Use .index.to_list()[0] to get the scalar index from cuDF Index
        movie_index = gpu_merged_df[gpu_merged_df['primaryTitle'].str.lower() == movie_title.lower()].index.to_list()[0]
    except IndexError:
        return f"Movie with title '{movie_title}' not found."

    # Get the similarity scores for the given movie from the GPU matrix
    # Transfer the row to CPU for sorting if necessary, or use cupy for sorting
    # Let's use cupy for sorting on GPU
    similarity_scores = gpu_cosine_sim[movie_index]

    # Get the indices of the top N similar movies (excluding the movie itself)
    # Use cupy.argsort for GPU-accelerated sorting
    # We need to exclude the movie itself, which has a similarity score of 1.0 at its own index.
    # Get indices that would sort in descending order, then take the top num_similar + 1, exclude the first one (the movie itself)
    top_similar_indices_gpu = cp.argsort(similarity_scores)[-num_similar-1:-1][::-1]

    # Convert the cupy array of indices to a pandas Index to use with .iloc on the cuDF DataFrame
    # cuDF iloc expects a pandas Index or a list of integers
    top_similar_indices_cpu = cp.asnumpy(top_similar_indices_gpu)
    top_similar_indices_pd = pd.Index(top_similar_indices_cpu)


    # Get the titles of the top N similar movies from the cuDF DataFrame
    # Use .iloc with the pandas Index
    similar_movies = gpu_merged_df['primaryTitle'].iloc[top_similar_indices_pd].to_pandas().tolist()


    return similar_movies

# Example usage (optional, for testing the function)
# Assuming 'Toy Story' is in the dataset
try:
    similar_movies_gpu_example = get_similar_movies_gpu("Toy Story", gpu_merged_df, gpu_cosine_sim, num_similar=5)
    print(f"Top 5 similar movies to Toy Story (GPU): {similar_movies_gpu_example}")
except Exception as e:
    print(f"An error occurred during GPU similarity calculation: {e}")
```

```python
import pandas as pd
import cudf
import cupy as cp
from cuml.feature_extraction.text import TfidfVectorizer as CumlTfidfVectorizer
from cuml.metrics.pairwise_kernels import cosine_similarity
from numba import cuda
import os
import gc # Import garbage collector

# List of files to download and extract
tsv_files = [
    'name.basics.tsv',
    'title.akas.tsv',
    'title.basics.tsv',
    'title.crew.tsv',
    'title.episode.tsv',
    'title.principals.tsv',
    'title.ratings.tsv'
]

# Check if all files exist, if not, download and extract
all_files_exist = all(os.path.exists(f) for f in tsv_files)

if not all_files_exist:
    print("Downloading and extracting data files...")
    for file_name in tsv_files:
        gz_file = file_name + '.gz'
        if not os.path.exists(file_name) and not os.path.exists(gz_file):
            print(f"Downloading {gz_file}...")
            !wget https://datasets.imdbws.com/{gz_file}
    # Extract all downloaded .gz files
    !gunzip -k /content/*.gz # Use -k to keep the .gz file
    # Clean up .gz files after extraction
    !rm -f /content/*.gz
    print("Download and extraction complete.")
else:
    print("Data files already exist.")

# Dictionary to store cuDF DataFrames
gpu_dfs = {}

# Load the necessary dataframes directly into cuDF DataFrames
print("Attempting to load data directly into cuDF...")
for file_name in tsv_files:
    try:
        print(f"Loading {file_name} into cuDF...")
        # Use dtype=str for columns that might contain mixed types or non-numeric strings
        if file_name in ['title.basics.tsv', 'title.episode.tsv', 'name.basics.tsv', 'title.principals.tsv', 'title.akas.tsv', 'title.crew.tsv']: # Specify all files that might have '\N' or mixed types
             gpu_dfs[file_name] = cudf.read_csv(file_name, sep='\t', dtype=str)
        else:
             gpu_dfs[file_name] = cudf.read_csv(file_name, sep='\t')
        print(f"Successfully loaded {file_name} into cuDF.")
    except Exception as e:
        print(f"Error loading {file_name} into cuDF: {e}")
        print(f"Skipping {file_name} for now or will attempt pandas fallback if necessary for merges.")
        gpu_dfs[file_name] = None # Mark as failed to load

# Perform merges using cuDF DataFrames that were successfully loaded
# Strategically merge, starting with a large central dataframe like title.basics or title.principals
gpu_merged_df = None

if gpu_dfs.get('title.basics.tsv') is not None:
    gpu_merged_df = gpu_dfs['title.basics.tsv']
    print("Starting merge with title.basics.tsv")

    if gpu_dfs.get('title.ratings.tsv') is not None:
        print("Merging with title.ratings.tsv")
        gpu_merged_df = gpu_merged_df.merge(gpu_dfs['title.ratings.tsv'], on='tconst', how='inner')
        del gpu_dfs['title.ratings.tsv'] # Free up memory
        gc.collect()


    if gpu_dfs.get('title.crew.tsv') is not None:
        print("Merging with title.crew.tsv")
        gpu_merged_df = gpu_merged_df.merge(gpu_dfs['title.crew.tsv'], on='tconst', how='inner')
        del gpu_dfs['title.crew.tsv'] # Free up memory
        gc.collect()

    # Principals is very large, merge carefully
    if gpu_dfs.get('title.principals.tsv') is not None:
         print("Merging with title.principals.tsv")
         # Inner merge on tconst to keep only titles with principal information
         gpu_merged_df = gpu_merged_df.merge(gpu_dfs['title.principals.tsv'], on='tconst', how='inner')
         del gpu_dfs['title.principals.tsv'] # Free up memory
         gc.collect()

    # Names is also large, merge after principals if needed for name details
    if gpu_dfs.get('name.basics.tsv') is not None:
        print("Merging with name.basics.tsv")
        # Merge on nconst to get name details for principals
        if 'nconst' in gpu_merged_df.columns:
             gpu_merged_df = gpu_merged_df.merge(gpu_dfs['name.basics.tsv'], on='nconst', how='inner')
             del gpu_dfs['name.basics.tsv'] # Free up memory
             gc.collect()
        else:
             print("Skipping merge with name.basics.tsv as 'nconst' column is not available after previous merges.")


    if gpu_dfs.get('title.episode.tsv') is not None:
        print("Merging with title.episode.tsv")
        gpu_merged_df = gpu_merged_df.merge(gpu_dfs['title.episode.tsv'], on='tconst', how='left')
        del gpu_dfs['title.episode.tsv'] # Free up memory
        gc.collect()

    if gpu_dfs.get('title.akas.tsv') is not None:
        print("Merging with title.akas.tsv")
        # Rename 'titleId' to 'tconst' before merging
        gpu_dfs['title.akas.tsv'] = gpu_dfs['title.akas.tsv'].rename(columns={'titleId': 'tconst'})
        gpu_merged_df = gpu_merged_df.merge(gpu_dfs['title.akas.tsv'], on='tconst', how='left')
        del gpu_dfs['title.akas.tsv'] # Free up memory
        gc.collect()

elif gpu_dfs.get('title.principals.tsv') is not None:
    # If basics loading failed, start with principals if available
     gpu_merged_df = gpu_dfs['title.principals.tsv']
     print("Starting merge with title.principals.tsv")
     # Attempt to merge other available dataframes with principals

     if gpu_dfs.get('title.basics.tsv') is not None:
        print("Merging with title.basics.tsv")
        gpu_merged_df = gpu_merged_df.merge(gpu_dfs['title.basics.tsv'], on='tconst', how='inner')
        del gpu_dfs['title.basics.tsv'] # Free up memory
        gc.collect()

     if gpu_dfs.get('title.ratings.tsv') is not None:
        print("Merging with title.ratings.tsv")
        gpu_merged_df = gpu_merged_df.merge(gpu_dfs['title.ratings.tsv'], on='tconst', how='inner')
        del gpu_dfs['title.ratings.tsv'] # Free up memory
        gc.collect()

     if gpu_dfs.get('title.crew.tsv') is not None:
        print("Merging with title.crew.tsv")
        gpu_merged_df = gpu_merged_df.merge(gpu_dfs['title.crew.tsv'], on='tconst', how='inner')
        del gpu_dfs['title.crew.tsv'] # Free up memory
        gc.collect()

     if gpu_dfs.get('name.basics.tsv') is not None:
        print("Merging with name.basics.tsv")
        if 'nconst' in gpu_merged_df.columns:
            gpu_merged_df = gpu_merged_df.merge(gpu_dfs['name.basics.tsv'], on='nconst', how='inner')
            del gpu_dfs['name.basics.tsv'] # Free up memory
            gc.collect()
        else:
             print("Skipping merge with name.basics.tsv as 'nconst' column is not available after previous merges.")

     if gpu_dfs.get('title.episode.tsv') is not None:
        print("Merging with title.episode.tsv")
        gpu_merged_df = gpu_merged_df.merge(gpu_dfs['title.episode.tsv'], on='tconst', how='left')
        del gpu_dfs['title.episode.tsv'] # Free up memory
        gc.collect()

     if gpu_dfs.get('title.akas.tsv') is not None:
        print("Merging with title.akas.tsv")
        gpu_dfs['title.akas.tsv'] = gpu_dfs['title.akas.tsv'].rename(columns={'titleId': 'tconst'})
        gpu_merged_df = gpu_merged_df.merge(gpu_dfs['title.akas.tsv'], on='tconst', how='left')
        del gpu_dfs['title.akas.tsv'] # Free up memory
        gc.collect()

else:
    print("Could not load essential dataframes (title.basics.tsv or title.principals.tsv) into cuDF. Cannot proceed with merging.")
    # Optionally raise an error or set a flag to indicate failure
    gpu_merged_df = None # Ensure merged_df is None if loading failed


# If merged_df was successfully created, proceed with cleaning and feature engineering
if gpu_merged_df is not None:
    print("\nMerging complete. Proceeding with data cleaning and feature engineering.")

    # Redo the data cleaning steps on the cuDF DataFrame
    # Convert to appropriate dtypes to save memory on GPU
    for col in ['runtimeMinutes', 'startYear', 'endYear']:
        if col in gpu_merged_df.columns:
            # Replace '\N' with None or a suitable placeholder before converting
            # Use .str.replace for string replacement in cuDF without regex
            gpu_merged_df[col] = gpu_merged_df[col].astype(str).str.replace('\\N', '', regex=False)
            # Attempt to convert to numeric, invalid parsing will result in NaN
            # cuDF's to_numeric can handle this similar to pandas errors='coerce'
            gpu_merged_df[col] = cudf.to_numeric(gpu_merged_df[col], errors='coerce').astype('float32')


    # Define and apply the feature combination function
    similarity_columns = [
        'genres',
        'directors',
        'writers',
        'primaryProfession',
        'category',
        'job',
        'primaryTitle',
        'originalTitle'
    ]

    # Fill missing values in cuDF DataFrame and ensure string type for concatenation
    for col in similarity_columns:
        if col in gpu_merged_df.columns:
            # Replace potential '\N' characters that might still be in string columns
            gpu_merged_df[col] = gpu_merged_df[col].astype(str).str.replace('\\N', '', regex=False).fillna('')
        else:
            # If a column is missing, add an empty string column for consistency
            gpu_merged_df[col] = ''
            print(f"Warning: Column '{col}' not found in gpu_merged_df during feature combination. Added as empty string.")


    # Combine features in cuDF using string concatenation
    print("Combining features...")
    # Use cuDF's string concatenation
    combined_features_series = gpu_merged_df[similarity_columns[0]].astype(str)
    for col in similarity_columns[1:]:
        combined_features_series = combined_features_series + ' ' + gpu_merged_df[col].astype(str)

    gpu_merged_df['combined_features'] = combined_features_series
    print("Feature combination complete.")

    # Instantiate cuML TfidfVectorizer
    print("Performing TF-IDF vectorization...")
    gpu_tfidf_vectorizer = CumlTfidfVectorizer(stop_words='english', max_features=10000)

    # Fit and transform the 'combined_features' column using the cuML TfidfVectorizer
    gpu_tfidf_matrix = gpu_tfidf_vectorizer.fit_transform(gpu_merged_df['combined_features'])
    print("TF-IDF vectorization complete.")

    # Calculate the cosine similarity matrix using the GPU TF-IDF matrix
    print("Calculating cosine similarity matrix...")
    # REMOVED: This was causing MemoryError. We will calculate similarity on demand.
    # gpu_cosine_sim = cosine_similarity(gpu_tfidf_matrix, gpu_tfidf_matrix)
    # print("Cosine similarity calculation complete.")

    # Display the shape of the resulting GPU TF-IDF matrix
    print("\nShape of the GPU TF-IDF matrix:")
    display(gpu_tfidf_matrix.shape)
    # Removed display of gpu_cosine_sim shape as it's no longer calculated fully
    # print("\nShape of the GPU cosine similarity matrix:")
    # display(gpu_cosine_sim.shape)

    # Display the head and info of the gpu_merged_df
    print("\nHead of the gpu_merged_df:")
    display(gpu_merged_df.head())

    print("\nInfo of the gpu_merged_df:")
    display(gpu_merged_df.info())

    # Redefine the GPU similarity function to use the new cuDF DataFrame and similarity matrix
    # The function will now calculate similarity on demand
    def get_similar_movies_gpu(movie_title, gpu_merged_df, gpu_tfidf_matrix, num_similar=5):
        """
        Finds the top N most similar movies to a given movie title using GPU.

        Args:
            movie_title: The title of the movie.
            gpu_merged_df: The cuDF DataFrame containing movie data.
            gpu_tfidf_matrix: The GPU-calculated TF-IDF matrix.
            num_similar: The number of similar movies to return (default is 5).

        Returns:
            A tuple containing a list of titles of the top N similar movies and the GPU execution time in milliseconds,
            or a tuple with an error message and None for time if the movie title is not found.
        """
        # Get the index of the movie title from the cuDF DataFrame
        try:
            # Use .index.to_list()[0] to get the scalar index from cuDF Index
            # Ensure the primaryTitle column is a string type for case-insensitive comparison
            movie_index = gpu_merged_df[gpu_merged_df['primaryTitle'].astype(str).str.lower() == movie_title.lower()].index.to_list()[0]
        except IndexError:
            return f"Movie with title '{movie_title}' not found.", None

        # Create CUDA events for timing
        start_event = cuda.Event()
        end_event = cuda.Event()

        # Start recording the GPU time
        start_event.record()

        # Get the TF-IDF vector for the given movie
        movie_vector = gpu_tfidf_matrix[movie_index]

        # Calculate the cosine similarity scores between the movie vector and all other movie vectors on demand
        # Corrected the cosine_similarity call to take two arguments
        similarity_scores = cosine_similarity(movie_vector, gpu_tfidf_matrix).flatten()

        # Get the indices of the top N similar movies (excluding the movie itself)
        # Use cupy.argsort for GPU-accelerated sorting
        # We need to exclude the movie itself, which has a similarity score of 1.0 at its own index.
        # Get indices that would sort in descending order, then take the top num_similar + 1, exclude the first one (the movie itself)
        top_similar_indices_gpu = cp.argsort(similarity_scores)[-num_similar-1:-1][::-1]

        # End recording the GPU time
        end_event.record()

        # Wait for the events to complete
        end_event.synchronize()

        # Calculate the elapsed time in milliseconds
        gpu_time_ms = cuda.event_elapsed_time(start_event, end_event)

        # Convert the cupy array of indices to a pandas Index to use with .iloc on the cuDF DataFrame
        # cuDF iloc expects a pandas Index or a list of integers
        top_similar_indices_cpu = cp.asnumpy(top_similar_indices_gpu)
        top_similar_indices_pd = pd.Index(top_similar_indices_cpu)

        # Get the titles of the top N similar movies from the cuDF DataFrame
        # Use .iloc with the pandas Index and convert to pandas Series before to_list()
        # CORRECTED: Use .to_arrow().to_pylist() instead of .to_pandas().tolist()
        similar_movies = gpu_merged_df['primaryTitle'].iloc[top_similar_indices_pd].to_arrow().to_pylist()

        return similar_movies, gpu_time_ms

    # Example usage
    movie_to_find = "Toy Story"
    try:
        # Pass gpu_tfidf_matrix instead of gpu_cosine_sim
        similar_movies_gpu_example, gpu_execution_time = get_similar_movies_gpu(movie_to_find, gpu_merged_df, gpu_tfidf_matrix, num_similar=5)

        if isinstance(similar_movies_gpu_example, list):
            print(f"Top 5 similar movies to {movie_to_find} (GPU):")
            for movie in similar_movies_gpu_example:
                print(movie)
            print(f"GPU execution time for similarity calculation: {gpu_execution_time:.2f} ms")
        else:
            print(similar_movies_gpu_example)

    except Exception as e:
        print(f"An error occurred during GPU similarity calculation: {e}")
else:
    print("\nFailed to create gpu_merged_df. Cannot proceed with similarity calculation.")
```
